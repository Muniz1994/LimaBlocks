```python
from GeomTools.CheckModel import CheckModel
import trimesh
import ifcopenshell
import numpy as np
from GeomTools.GeometricFunctions import get_depth
```

```python
# load model
my_model = CheckModel("TestFile/BuildingPaper.V17.ifc")
```

```python
building = my_model.select('.IfcBuilding')[0]
```

```python
walls = my_model.select('@ #%s & .IfcWall'%str(building.ifc.GlobalId))

under_walls = []

for wall in walls:
    # Access the y-coordinates (second column, index 1)
    y_coordinates = wall.mesh.vertices[:, 2]

    # Check if any y-coordinate is lower than 0
    if not np.any(y_coordinates < 0):
        # If so, remove the wall from the list
        under_walls.append(wall)

empena = sum([w.mesh for w in under_walls])
```
